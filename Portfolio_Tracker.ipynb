```python
# Hardcoded stock prices dictionary
STOCK_PRICES = {
    "AAPL": 180,
    "TSLA": 250,
    "GOOGL": 140,
    "MSFT": 330,
    "AMZN": 145,
    "NVDA": 460,
    "META": 300,
    "NFLX": 450
}


def display_available_stocks():
    """Display available stocks and their prices."""
    print("\n" + "=" * 50)
    print("AVAILABLE STOCKS")
    print("=" * 50)
    for symbol, price in STOCK_PRICES.items():
        print(f"{symbol}: ${price}")
    print("=" * 50 + "\n")


def get_user_input():
    """Get stock name and quantity from user."""
    portfolio = []

    print("Enter your stock holdings (type 'done' when finished):")
    print("-" * 50)

    while True:
        stock_name = input("\nEnter stock symbol (or 'done' to finish): ").strip().upper()

        if stock_name == 'DONE':
            break

        if stock_name not in STOCK_PRICES:
            print(f"Error: '{stock_name}' is not a valid stock symbol.")
            print("Available symbols:", ', '.join(STOCK_PRICES.keys()))
            continue

        try:
            quantity = int(input(f"Enter quantity for {stock_name}: "))
            if quantity <= 0:
                print("Error: Quantity must be a positive number.")
                continue
        except ValueError:
            print("Error: Please enter a valid number for quantity.")
            continue

        portfolio.append({
            'symbol': stock_name,
            'quantity': quantity,
            'price': STOCK_PRICES[stock_name]
        })

        print(f"✓ Added {quantity} shares of {stock_name} at ${STOCK_PRICES[stock_name]} each")

    return portfolio


def calculate_investment(portfolio):
    """Calculate total investment value."""
    total = 0
    holdings = []

    for stock in portfolio:
        value = stock['quantity'] * stock['price']
        total += value
        holdings.append({
            'symbol': stock['symbol'],
            'quantity': stock['quantity'],
            'price': stock['price'],
            'value': value
        })

    return holdings, total


def display_portfolio(holdings, total):
    """Display the portfolio summary."""
    print("\n" + "=" * 70)
    print("PORTFOLIO SUMMARY")
    print("=" * 70)
    print(f"{'Symbol':<10} {'Quantity':<10} {'Price':<10} {'Value':<15}")
    print("-" * 70)

    for stock in holdings:
        print(f"{stock['symbol']:<10} {stock['quantity']:<10} ${stock['price']:<9} ${stock['value']:<14,.2f}")

    print("-" * 70)
    print(f"{'TOTAL INVESTMENT:':<45} ${total:,.2f}")
    print("=" * 70)


def save_to_txt(holdings, total, filename="portfolio_summary.txt"):
    """Save portfolio summary to a text file."""
    with open(filename, 'w') as f:
        f.write("PORTFOLIO SUMMARY\n")
        f.write("=" * 70 + "\n\n")
        f.write(f"{'Symbol':<10} {'Quantity':<10} {'Price':<10} {'Value':<15}\n")
        f.write("-" * 70 + "\n")

        for stock in holdings:
            f.write(f"{stock['symbol']:<10} {stock['quantity']:<10} ${stock['price']:<9} ${stock['value']:<14,.2f}\n")

        f.write("-" * 70 + "\n")
        f.write(f"{'TOTAL INVESTMENT:':<45} ${total:,.2f}\n")
        f.write("=" * 70 + "\n")

    print(f"\n✓ Portfolio saved to '{filename}'")


def save_to_csv(holdings, total, filename="portfolio_summary.csv"):
    """Save portfolio summary to a CSV file."""
    with open(filename, 'w') as f:
        f.write("Symbol,Quantity,Price,Value\n")
        for stock in holdings:
            f.write(f"{stock['symbol']},{stock['quantity']},{stock['price']},{stock['value']}\n")
        f.write(f"TOTAL,,,,${total:,.2f}\n")

    print(f"✓ Portfolio saved to '{filename}'")


def main():
    """Main function to run the stock portfolio tracker."""
    print("\n" + "=" * 50)
    print("STOCK PORTFOLIO TRACKER")
    print("=" * 50)

    # Display available stocks
    display_available_stocks()

    # Get user input
    portfolio = get_user_input()

    if not portfolio:
        print("\nNo stocks entered. Exiting...")
        return

    # Calculate investment
    holdings, total = calculate_investment(portfolio)

    # Display portfolio
    display_portfolio(holdings, total)

    # Save options
    print("\nSave Options:")
    print("1. Save to TXT file")
    print("2. Save to CSV file")
    print("3. Skip saving")

    choice = input("\nEnter your choice (1/2/3): ").strip()

    if choice == '1':
        filename = input("Enter filename (default: portfolio_summary.txt): ").strip()
        if not filename:
            filename = "portfolio_summary.txt"
        save_to_txt(holdings, total, filename)
    elif choice == '2':
        filename = input("Enter filename (default: portfolio_summary.csv): ").strip()
        if not filename:
            filename = "portfolio_summary.csv"
        save_to_csv(holdings, total, filename)
    else:
        print("\nSkipping save option.")

    print("\nThank you for using Stock Portfolio Tracker!")


if __name__ == "__main__":
    main()
```
